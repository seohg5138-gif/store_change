```python
import pandas as pd
df = pd.read_parquet("df_all_slotted.parquet")

top_locations = (
    df.groupby(["PNU19", "층_최종"])["슬롯번호"].max()
    .sort_values(ascending=False)
    .head(20)
)
print(top_locations)

# 이 중 하나를 골라서 신뢰가능여부 확인
pnu, floor = top_locations.index[0]
sample = df[(df["PNU19"] == pnu) & (df["층_최종"] == floor)]
print(sample["층정보_신뢰가능여부"].value_counts())
```

```python
import pandas as pd
df = pd.read_parquet("df_all_slotted.parquet")

nan_floor_rows = df[df["층_최종"] == "nan"]
print(f"층_최종='nan'인 행 수: {len(nan_floor_rows):,} / 전체 {len(df):,} ({len(nan_floor_rows)/len(df)*100:.2f}%)")
print(nan_floor_rows["층정보_신뢰가능여부"].value_counts())
```

```python
verified = df[df["층정보_신뢰가능여부"] == "검증완료"]
max_slot = verified.groupby(["PNU19", "층_최종"])["슬롯번호"].max()
print(max_slot.value_counts().sort_index(ascending=False).head(10))
```

```python
# 필터 없이 vs 검증완료만 걸렀을 때, 최대 슬롯 분포가 다른지 비교
all_max = df.groupby(["PNU19","층_최종"])["슬롯번호"].max()
verified_max = df[df["층정보_신뢰가능여부"]=="검증완료"].groupby(["PNU19","층_최종"])["슬롯번호"].max()

print("[전체]")
print(all_max.sort_values(ascending=False).head(10))
print("\n[검증완료만]")
print(verified_max.sort_values(ascending=False).head(10))
```

```python
import pandas as pd
df = pd.read_parquet("shop_period_raw.parquet")

nan_floor = df[df["층정보_신뢰가능여부"] == "검증불가_층정보없음"]
print("연도별 분포:")
print((nan_floor["period"] // 100).value_counts().sort_index())

print("\n전체 대비 결측 비율 (연도별):")
total_by_year = (df["period"] // 100).value_counts().sort_index()
nan_by_year = (nan_floor["period"] // 100).value_counts().sort_index()
print((nan_by_year / total_by_year * 100).round(1))
```
